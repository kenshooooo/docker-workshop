```python
import pandas as pd
```

```python
df = pd.read_parquet("green_tripdata_2025-11.parquet")
zones = pd.read_csv("taxi_zone_lookup.csv")
```

```python
##Q3
start = pd.Timestamp("2025-11-01")
end = pd.Timestamp("2025-12-01")

count = (
    df[
        (df["lpep_pickup_datetime"] >= start) &
        (df["lpep_pickup_datetime"] < end) &
        (df["trip_distance"] <= 1)
    ]
    .shape[0]
)

count
```

```python
import pandas as pd

start = pd.Timestamp("2025-11-18")
end = pd.Timestamp("2025-11-19")

df_day = df[
    (df["lpep_pickup_datetime"] >= start) &
    (df["lpep_pickup_datetime"] < end)
]
```

```python
agg = (
    df_day
    .groupby("PULocationID", as_index=False)["total_amount"]
    .sum()
)
```

```python
agg = agg.merge(
    zones,
    left_on="PULocationID",
    right_on="LocationID",
    how="left"
)
```

```python
result = agg.sort_values("total_amount", ascending=False).iloc[0]

result["Zone"], result["Borough"], result["total_amount"]
```

```python
start = pd.Timestamp("2025-11-01")
end = pd.Timestamp("2025-12-01")

df_nov = df[
    (df["lpep_pickup_datetime"] >= start) &
    (df["lpep_pickup_datetime"] < end)
]

df_nov = df_nov.merge(
    zones[["LocationID", "Zone"]],
    left_on="PULocationID",
    right_on="LocationID",
    how="left"
).rename(columns={"Zone": "PUZone"})


df_ehn = df_nov[df_nov["PUZone"] == "East Harlem North"]

df_ehn = df_ehn.merge(
    zones[["LocationID", "Zone"]],
    left_on="DOLocationID",
    right_on="LocationID",
    how="left"
).rename(columns={"Zone": "DOZone"})

tips_by_zone = (
    df_ehn
    .groupby("DOZone", as_index=False)["tip_amount"]
    .sum()
)


result = tips_by_zone.sort_values("tip_amount", ascending=False).iloc[0]

result["DOZone"], result["tip_amount"]

```

```python

tips_by_zone_sorted = tips_by_zone.sort_values(
    by="tip_amount",
    ascending=False
)

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None,
    "display.max_colwidth", None
):
    print(tips_by_zone_sorted)
```
